```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def load_img():
    blank_img =np.zeros((600,600))
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img,text='SANA',org=(90,300), fontFace=font,fontScale= 5,color=(255,255,255),thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img,cmap='gray')
    plt.show()
```

```python
img = load_img()
```

```python
noise = np.random.randint(0, 256, (600, 600), dtype=np.uint8)
```

```python
noise = noise*255
```

```python
noise
```

```python
noise_img = noise+img
```

```python
display_img(noise_img) 
```

```python
kernel = np.ones((5, 5), np.uint8)
```

```python
kernel
```

```python
opening = cv2.morphologyEx(noise_img, cv2.MORPH_OPEN, kernel)
```

```python
display_img(opening)   # Expected Output 
```
